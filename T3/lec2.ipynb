```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

```python
with open('vishal_new.html','r',encoding="utf-8") as f:
    res = f.read()
sp = BeautifulSoup(res,'html.parser')
```

```python
d={'mobilename':[],'star':[],'rating':[],'review':[],'display':[],'camera':[],'new_price':[],'old_price':[],'discount':[]}
```

```python
x = sp.find_all('div',class_='_3pLy-c row')
```

```python
import numpy as np
for i in x:
    d['mobilename'].append(i.find("div",class_="_4rR01T").text.strip())
    d['rating'].append(i.find("span",class_="_2_R_DZ").text.split("&")[0][:-9].replace(",",""))
    # d['review'].append(i.find("div",class_="_2_R_DZ").text.split("&")[1][0:-8].replace(",",""))
    d['star'].append(float(i.find("div",class_="_3LWZlK").text.strip()))
    try: 
        d['old_price'].append(i.find("div",class_="_3I9_wc _27UcVY").text.strip())
    except:
        d['old_price'].append(np.nan)
      
    try:
        d['new_price'].append(i.find("div",class_="_30jeq3 _1_WHN1").text.strip())
    except:
        d['new_price'].append(np.nan)

    try:
        d['discount'].append((i.find("div",class_="_3Ay6Sb").text.strip()))
    except:
        d['discount'].append(np.nan)
print(d)
```

```python
with open("List of state and union territory capitals in India - Wikipedia (5_28_2025 8：46：50 AM).html",'r',encoding="utf-8") as f:
    script = f.read()
sp = BeautifulSoup(script,"html.parser")
```

```python
table = sp.find('table',class_="wikitable sortable jquery-tablesorter")
body = table.find('tbody')
box = table.find_all("tr")
d = {'State':[],'AEC':[],'LC':[],'JC':[],'YOE':[],'FC':[]}
box
```

```python
d = {'State':[],'AEC':[],'LC':[],'JC':[],'YOE':[],'FC':[]}
for i in box:
    l = i.find_all('td')

    d['State'].append(l[0].text.strip())
    d['AEC'].append(l[1].text.strip())
    d['LC'].append(l[2].text.strip())
    d['JC'].append(l[3].text.strip())
    d['YOE'].append(l[4].text.strip())
    d['FC'].append(l[5].text.strip())

d
```

```python
pd.DataFrame(d)
```
